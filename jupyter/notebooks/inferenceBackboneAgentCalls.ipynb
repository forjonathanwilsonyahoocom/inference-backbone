```python
import requests

url = "http://worker-agent:8000/run"

task_description = """
in the workspace there is a project

./inference-backone

DISCOVERY AND CREATION MISSION

we have reached a milestone! we added some metrics to allow better observation!
i also have refactored the worker-agent code (that you are touching with tool calls), 
and it seems that i have made some human syntax errors and missed some imports, so please use all the tools so that i find the problems!

would you please inspect the project and see what you find while i watch them?

you have already helped on several of these iterations, and we may have found all the bugs,
"""
payload = {
    "request": task_description,
    "verbose": False,
    "max_iterations":100
}

try:
    response = requests.post(
        url,
        json=payload,  # Serializes the dict and sets Content-Type: application/json
        timeout=1000,
    )
    display(Audio('./freesound_community-chime-74910.mp3', autoplay=True))
    response.raise_for_status()

    # If the endpoint returns JSON:
    result = response.json()
    display(Markdown(result["final_response"]))

except requests.exceptions.RequestException as error:
    print(f"HTTP request failed: {error}")
```
