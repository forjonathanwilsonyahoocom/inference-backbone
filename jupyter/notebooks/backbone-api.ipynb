```python
import requests
file_content = []

for id in result["evidence_ids"]:
    url = f"http://backbone-api:8000/file/claim/{id}"
    
    response = requests.get(
        url,
        timeout=1000,
    )
    thing = response.json()
    
        
    url = "http://backbone-api:8000/search/evidence"
    
    payload = {
      "query": thing["content"],
      "limit" : 20,
      "execution_id" : eid,
    }
    response = requests.post(
        url,
        json=payload,  # Serializes the dict and sets Content-Type: application/json
        timeout=1000,
    )
    search_hits = response.json()

    
    file_content.append({"source": thing,
                        "related" : search_hits})

```

```python
nearby = set()
for x in file_content:
    if x["source"]["importance"] > 0.5:
        print(x["source"]["importance"], x["source"]["claim_id"])
        for y in x["related"]:
            nearby.add( y["properties"]["evidence_id"])
           # print(y["distance"], y["properties"]["evidence_id"])
    else:
        print(x["source"]["importance"] , "skipped")
len(nearby)
```

```python
from IPython.display import display, JSON
import requests

url = "http://backbone-api:8000/search/info"

payload = {
  "query": "list_files executed on vr-mbe directory, confirming presence of src/ and tests/ direc",
  "limit" : 10,
 # "execution_id" : "fa6055ab-5acc-4d04-bda8-098b0a7528e0",
# "instance_number" : 2
}

try:
    response = requests.post(
        url,
        json=payload,  # Serializes the dict and sets Content-Type: application/json
        timeout=1000,
    )

    response.raise_for_status()

    # If the endpoint returns JSON:
    result = response.json()

except requests.exceptions.RequestException as error:
    print(f"HTTP request failed: {error}")

result.sort(key=lambda x: x['properties']['chunk_index'])

for r in result:
    print("chunk_index", r['properties']['chunk_index'], "distance", r["distance"])
display(JSON(result, expanded=True))
```
