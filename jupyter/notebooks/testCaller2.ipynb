```python
import requests

url = "http://backbone-api:8000/list/evidence/65c48c14-ad2e-4011-ae11-ddc51d92a3e4"

payload = {
  "query": "list_files",
  "limit" : 500
}

try:
    response = requests.get(
        url,
        timeout=1000,
    )

    response.raise_for_status()

    # If the endpoint returns JSON:
    result = response.json()


except requests.exceptions.RequestException as error:
    print(f"HTTP request failed: {error}")
    

display(JSON(result, expanded=True))
```

```python
import requests

url = "http://backbone-api:8000/search/evidence"

payload = {
  "query": "brown fox",
  "limit" : 500
}

try:
    response = requests.post(
        url,
        json=payload,  # Serializes the dict and sets Content-Type: application/json
        timeout=1000,
    )

    response.raise_for_status()

    # If the endpoint returns JSON:
    result = response.json()

except requests.exceptions.RequestException as error:
    print(f"HTTP request failed: {error}")


display(JSON(result, expanded=False))
```

```python
import requests

url = "http://web-tools:8000/fetch"

payload = {
  "url": "https://en.wikipedia.org/wiki/Fox"
}

try:
    response = requests.post(
        url,
        json=payload,  # Serializes the dict and sets Content-Type: application/json
        timeout=1000,
    )

    response.raise_for_status()

    # If the endpoint returns JSON:
    result = response.json()

except requests.exceptions.RequestException as error:
    print(f"HTTP request failed: {error}")


display(JSON(result, expanded=False))
```

```python
import requests, json, os

repo = "inference-backbone"
url = f"http://graphdb:7200/repositories/{repo}"
sparql = """
PREFIX ex: <http://mindbodyengineer.com/claim>
SELECT (COUNT(?c) AS ?count)
WHERE { ?doc a ex:Claim }
"""

resp = requests.post(url, data={"query": sparql},
                     headers={"Accept": "application/sparql-results+json"})

display(JSON(resp.json(), expanded=True))
```
